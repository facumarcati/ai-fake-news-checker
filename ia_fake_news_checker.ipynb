```python
# ======================================
# AI Fake News Checker - Entrega Final
# ======================================

import openai
from openai import OpenAI
from IPython.display import Image, display

# Inicializar cliente
client = OpenAI(api_key="")  # Reemplazar con la API key

# ============================
# Introducción (Markdown)
# ============================

"""
# AI Fake News Checker - Notebook

Este notebook muestra la implementación de la POC:
- **Texto→Texto**: verificación de noticias falsas con prompts básicos y refinados.
- **Texto→Imagen**: generación de imágenes ilustrativas con herramientas externas.

Se aplican técnicas de **Fast Prompting** para optimizar resultados.
"""

# ============================
# Ejemplo 1 - Noticia falsa
# ============================

fake_new = """Científicos confirman que beber 10 litros de agua con azúcar cura todas las enfermedades."""

prompt_basico = f"¿Es verdadera esta noticia?\n\n{fake_new}"

respuesta_basica = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": prompt_basico}]
)

print("🔹 Respuesta básica:\n")
print(respuesta_basica.choices[0].message.content)

# ============================
# Fast Prompting
# ============================

prompt_fast = f"""Actúa como un verificador de hechos. Analiza la siguiente noticia: {fake_new}
- Detecta inconsistencias
- Menciona fuentes confiables
- Responde en un máximo de 3 puntos"""

respuesta_fast = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": prompt_fast}]
)

print("🔹 Respuesta con Fast Prompting:\n")
print(respuesta_fast.choices[0].message.content)

# ============================
# Ejemplos adicionales
# ============================

noticias = [
    "El telescopio James Webb confirmó la existencia de agua en un exoplaneta.",
    "Si dormís menos de 2 horas al día, tu cerebro funciona mejor.",
    "La vacuna contra la gripe contiene chips de rastreo."
]

for n in noticias:
    print("\n===================================")
    print(f"Noticia: {n}")
    
    p_fast = f"""Actúa como un verificador de hechos. Analiza la siguiente noticia: {n}
    - Detecta inconsistencias
    - Menciona fuentes confiables
    - Responde en un máximo de 3 puntos"""
    
    r = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": p_fast}]
    )
    print(r.choices[0].message.content)

# ============================
# Texto → Imagen
# ============================

"""
## Generación de imágenes

Para complementar el análisis textual, se generó una imagen ilustrativa con NightCafe / Stable Diffusion.
Prompt utilizado:
"Infografía digital que representa cómo se propaga una noticia falsa en redes sociales: teléfonos con titulares alarmistas, iconos de compartir, estilo flat, colores llamativos."
"""

# Mostrar la imagen generada (guardada en la carpeta images/)
display(Image("images/fake_news_infographic.png"))

# ============================
# Resultados y conclusiones
# ============================

"""
## Resultados
- El prompt básico devolvió respuestas genéricas.
- El Fast Prompting generó análisis más estructurados y útiles.
- Las imágenes generadas aportan un recurso visual para comunicar mejor.

## Conclusiones
- El Fast Prompting mejora significativamente la verificación automática de noticias.
- El proyecto combina texto→texto e imagen para crear una herramienta más clara y didáctica.
- Se cumplieron los objetivos de la entrega final.
"""
```
