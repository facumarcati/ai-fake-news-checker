```python
import openai import OpenAI
```

```python
client = OpenAI(api_key="")
```

> Importante: Para ejecutar este notebook necesitás una API Key de OpenAI.  
> Reemplazá `""` por tu propia clave.

```python
fake_new = """Científicos confirman que beber 10 litros de agua con azúcar cura todas las enfermedades."""
```

```python
prompt_basico = f"¿Es verdadera esta noticia?\n\n{fake_new}"
```

```python
# Prompt basico

respuesta_basica = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": prompt_basico}]
)

print("🔹 Respuesta básica:\n")
print(respuesta_basica.choices[0].message.content)
```

```python
# Prompt con Fast Prompting

prompt_fast = f"""Actúa como un verificador de hechos. Analiza la siguiente noticia: {fake_new}
- Detecta inconsistencias
- Menciona fuentes confiables
- Responde en un máximo de 3 puntos"""

respuesta_fast = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": prompt_fast}]
)

print("🔹 Respuesta con Fast Prompting:\n")
print(respuesta_fast.choices[0].message.content)
```
